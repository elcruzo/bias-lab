## 1. Algorithm Overview

**Contrastive Framing Attribution (CFA)** answers: *Which exact phrases push bias scores away from neutral, and by how much?*

### Mathematical Foundation

For each dimension $d \in \{ideology, factual, framing, tone, transparency\}$ and sentence $i$:

- **Baseline Score**: $f_d(x)$ - Original article score
- **Neutral Score**: $f_d(NR)$ - Neutral rewrite score  
- **Without Sentence**: $f_d(x \setminus i)$ - Score without sentence $i$
- **Contribution**: $C_{d,i} = [f_d(x) - f_d(NR)] - [f_d(x \setminus i) - f_d(NR)]$

Where:
- $f_d(x)$: Scoring function for dimension $d$ on text $x$
- $NR$: Neutral rewrite of the article
- $x \setminus i$: Article with sentence $i$ removed

**Interpretation**: Positive $C_{d,i}$ means sentence $i$ pushes the score away from neutral in dimension $d$.
